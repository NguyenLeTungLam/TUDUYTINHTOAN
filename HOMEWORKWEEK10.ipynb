Cho hai số nguyên dương, n và m (0 < n, m <= 20 ). .Bạn hãy in ra tất cả các chuỗi nhị phân có
n số 0 và m số 1. In theo thứ tự tăng dần trong từ điển

```python
def solve():
    try:
        line = input()
        parts = line.split()
        n = int(parts[0])
        m = int(parts[1])
    except (IOError, IndexError, ValueError):
        return

    total_length = n + m

    current_path = [''] * total_length

    def backtrack(position, count_0, count_1):

        if position == total_length:

            print(''.join(current_path))
            return

        if count_0 < n:
            current_path[position] = '0'
            backtrack(position + 1, count_0 + 1, count_1)

        if count_1 < m:
            current_path[position] = '1'
            backtrack(position + 1, count_0, count_1 + 1)


    backtrack(0, 0, 0)
solve()
```

Viết hàm tìm dãy con tăng nghiêm ngặt dài nhất trong dãy đầu vào từ bàn phím 

```python
def f():
    a = [int(x) for x in input("Nhập dãy số: ").split()]
    if not a: return

    L = [1] * len(a)

    for i in range(len(a)):
        for j in range(i):      
            if a[j] < a[i]:  
                L[i] = max(L[i], L[j] + 1) 


    print("Độ dài dãy con tăng dài nhất là:", max(L))

f()

```

Giả sử mật khẩu mạnh là một chuỗi kí tự dài ít nhất 8 kí tự gồm ít nhất 1 chữ cái thường, 1 chữ
cái hoa, 1 chữ số và 1 trong các kí tự đặc biệt !, @, #, $. Nhập vào một chuỗi kĩ tự, kiểm tra xem
nó có phải là mật khẩu mạnh không.

```python
def kiem_tra_mat_khau():
    password = input("Nhập mật khẩu cần kiểm tra: ")
    if len(password) < 8:
        print("Mật khẩu YẾU (Phải dài ít nhất 8 kí tự)")
        return

    co_chu_thuong = False
    co_chu_hoa = False
    co_chu_so = False
    co_ki_tu_dac_biet = False

    ki_tu_dac_biet_hop_le = "!" + "@" + "#" + "$"  


    for char in password:
        if char.islower():
            co_chu_thuong = True
        elif char.isupper():
            co_chu_hoa = True
        elif char.isdigit():
            co_chu_so = True
        elif char in ki_tu_dac_biet_hop_le:
            co_ki_tu_dac_biet = True

    if co_chu_thuong and co_chu_hoa and co_chu_so and co_ki_tu_dac_biet:
        print("Mật khẩu MẠNH")
    else:
        print("Mật khẩu YẾU (Thiếu chữ hoa, chữ thường, số hoặc kí tự đặc biệt !, @, #, $)")

kiem_tra_mat_khau()
```

Nén chuối kí tự bằng cách đếm số lần xuất hiện của các kí tự liên tiếp nhau và ghi lại ngay sau
kí tự đó. VD aaaaaafdhhhhh =>a6f1d1h5

```python
def nen_chuoi():

    s = input("Nhập chuỗi cần nén: ")
    if not s: 
        return

    ket_qua = []
    dem = 1  


    for i in range(len(s) - 1):
        if s[i] == s[i + 1]:

            dem += 1
        else:

            ket_qua.append(s[i] + str(dem))
            dem = 1  

    ket_qua.append(s[-1] + str(dem))


    print("Chuỗi sau khi nén:", "".join(ket_qua))

nen_chuoi()
```

Tìm phần tử lớn thứ k trong dãy n phần tử bằng thuật toán tốt nhất có thể – Tránh dùng sắp xếp
(dùng phân hoạch của quicksort) 


```python
def quickselect(a, k):
    pivot = a[0]
    lon_hon = [x for x in a if x > pivot]
    bang    = [x for x in a if x == pivot]
    nho_hon = [x for x in a if x < pivot]

    if k <= len(lon_hon):
        return quickselect(lon_hon, k)
    elif k <= len(lon_hon) + len(bang):
        return pivot  
    else:
        return quickselect(nho_hon, k - len(lon_hon) - len(bang))

def main():
    a = [int(x) for x in input("Nhập dãy số: ").split()]
    k = int(input("Nhập k: "))
    
    if 1 <= k <= len(a):
        print(f"Phần tử lớn thứ {k} là:", quickselect(a, k))
    else:
        print("k không hợp lệ!")

main()
```

Loại bỏ khỏi danh sách các phần tử trùng lặp, giữ lại đúng 1 lần xuất hiện, in lại theo đúng thứ tự
xuất hiện 

```python
def xoa_trung_lap():
    a = [int(x) for x in input("Nhập dãy số: ").split()]
    
    ket_qua = []
    da_gap = set()
    for x in a:
        if x not in da_gap:
            ket_qua.append(x)
            da_gap.add(x) 
            
    print("Dãy số sau khi lọc trùng:", *ket_qua)

xoa_trung_lap()
```

Cho số nguyên rất lớn (string), và một số nguyên dương k. Tìm số có k chữ số lớn nhất có thể
từ số ban đầu bằng cách xóa bớt đi một số chữ số nào đó mà không thay đổi vị trí của bất kì
chữ số nào. (VD: 78934340999, k = 7 => 9440999 là số lớn nhất có 7 chữ số cần tìm) 

Hai chuỗi là anagram nếu chúng có cùng ký tự và số lần xuất hiện. viết CT kiểm tra xem 2 chuỗi
đầu vào có phải là anagram của nhau không 


```python
from collections import Counter

def check_anagram():

    s1 = input("Nhập chuỗi thứ nhất: ").strip()
    s2 = input("Nhập chuỗi thứ hai: ").strip()
    if Counter(s1) == Counter(s2):
        print("Hai chuỗi LÀ anagram của nhau.")
    else:
        print("Hai chuỗi KHÔNG PHẢI là anagram của nhau.")
check_anagram()
```

```python
def main():
    s = input("Nhập số nguyên lớn: ").strip()
    k = int(input("Nhập k: "))
    
    xoa = len(s) - k  
    stack = []
    
    for char in s:

        while stack and stack[-1] < char and xoa > 0:
            stack.pop()
            xoa -= 1
        stack.append(char)
    print("Số lớn nhất cần tìm là:", "".join(stack[:k]))

main()
```
