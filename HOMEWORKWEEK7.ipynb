```python
def count_names(s):
    count = 0
    words = s.split()
    
    for word in words:
        clean_word = "".join(filter(str.isalpha, word))
        if len(clean_word) >= 2 and clean_word.istitle():
            count += 1
            
    return count

xau_nhap_vao = input("Nhập xâu kí tự: ")
ket_qua = count_names(xau_nhap_vao)
print(ket_qua)
```

```python
def normalize(arr):
    if len(arr) > 1 and arr[0] > arr[-1]:
        return arr[::-1]
    return arr

def merge_sorted_arrays(arr1, arr2):
    a = normalize(arr1)
    b = normalize(arr2)
    
    result = []
    i = 0
    j = 0

    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
            
    result.extend(a[i:])
    result.extend(b[j:])   
    return result
print("Nhập mảng thứ nhất (ví dụ: 1 3 5 hoặc 5 3 1):")
mang1 = list(map(int, input().split()))
print("Nhập mảng thứ hai (ví dụ: 2 4 6 hoặc 6 4 2):")
mang2 = list(map(int, input().split()))
ket_qua = merge_sorted_arrays(mang1, mang2)
print("Mảng sau khi gộp và sắp xếp:")
print(ket_qua)
```

```python
Nhâ ̣p vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k)
```

```python

n, m, k = map(int, input().split())
matrix = []
for _ in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)
idx = k - 1
sum_row = sum(matrix[idx]) if 0 <= idx < n else 0

sum_col = sum(matrix[i][idx] for i in range(n)) if 0 <= idx < m else 0

print(f"Tong hang {k}: {sum_row}")
print(f"Tong cot {k}: {sum_col}")
```
