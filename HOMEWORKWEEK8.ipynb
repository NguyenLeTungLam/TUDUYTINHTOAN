Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.



Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng
dần, lẻ giảm dần; ; 


```python
lst = list(map(int , input().split()))
chan = []
le = []
for i in lst:
    if i % 2 == 0:
        chan.append(i)
    else:
        le.append(i)
chan.sort()
le.sort(reverse = True)
result = chan + le
print(result)
```


Cho N là ngày, T là tháng năm 2020, Tính xem đó là ngày thứ bao nhiêu
của năm đó nếu đó là ngảy hợp lệ


```python
def tinh_thu_cong(n, t): 
    ngay_trong_thang = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

    if t < 1 or t > 12 or n < 1 or n > ngay_trong_thang[t-1]:
        return "Ngày không hợp lệ"
    
    so_thu_tu = sum(ngay_trong_thang[:t-1]) + n
    return so_thu_tu
N = 15
T = 5
print(f"Ngày {N}/{T}/2020 là ngày thứ: {tinh_thu_cong(N, T)}")
```

Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các
số đó

```python
def giai_quyet_bai_toan(s):
    numbers = []
    current_num = ""
    for char in s:

        if '0' <= char <= '9' or char == '.':
            current_num += char
        else:

            if current_num:

                if current_num != ".":
                    numbers.append(float(current_num))
                current_num = ""

    if current_num and current_num != ".":
        numbers.append(float(current_num))
    
  
    if not numbers:
        return "Không có số nào trong văn bản."
    
    tong = 0
    for n in numbers:
        tong += n
        
    trung_binh = tong / len(numbers)
    
    return numbers, tong, trung_binh


text = "Gia 15.5 nghin, mua 3 cai het 46.5 nghin"
ds_so, t, tb = giai_quyet_bai_toan(text)
print("Các số tìm thấy:", ds_so)
print("Tổng:", t)
print("Trung bình:", tb)
```

Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python

s = input("Nhập các số (cách nhau bởi dấu cách): ")
words = s.split()
a = []
for x in words:
    a.append(int(x))

def tim_day_con_tang_dai_nhat(a):
    if not a:
        return []

    max_len = 1      
    current_len = 1   
    start_index = 0   
    
    current_start = 0 

    for i in range(1, len(a)):
        if a[i] > a[i-1]:
            current_len += 1
        else:

            if current_len > max_len:
                max_len = current_len
                start_index = current_start

            current_len = 1
            current_start = i


    if current_len > max_len:
        max_len = current_len
        start_index = current_start
    res = a[start_index : start_index + max_len]
    return res

# Thực thi và in kết quả
ket_qua = tim_day_con_tang_dai_nhat(a)
print("Dãy con tăng dần liên tục dài nhất đầu tiên là:")
for x in ket_qua:
    print(x, end=' ')
```
