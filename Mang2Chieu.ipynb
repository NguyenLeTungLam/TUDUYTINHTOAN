Bài 6: Cho ma trận nxm số nguyên, thêm vào một cột tại vị trí k.

```python
n , m = map(int , input().split())
mt = []
for i in range(n):
    row = list(map(int , input().split()))
    mt.append(row)
k = int(input())
cm = []
for i in range(n):
    val = int(input())
    cm.append(val)
for i in range(n):
    mt[i].insert(k , cm[i])
print(mt , end = " ")
```

Bài 7: Cho ma trận số nguyên nxm, xóa đi hàng k (0 <= k < n).

```python
n , m = map(int , input().split())
mt = []
for i in range(n):
    row = list(map(int, input().split()))
    mt.append(row)
k = int(input())
mt.pop(k)
for row in mt:
    print(*row)
```
