Viết chương trình nhập vào một tuple, in ra phần tử đầu, phần tử cuối, và
tuple đảo ngược


```python
tup = tuple(map(int , input().split()))
if len(tup) >0:
    sodau = tup[0]
    socuoi = tup[-1]
    daongc = tup[::-1]
print(f"so dau la: {sodau} , so cuoi la: {socuoi} , dao nguoc la: {daongc}")
```

Viết chương trình nhập vào một list các chuỗi, sau đó đếm số lần xuất hiện
của từng chuỗi và lưu vào dictionary.


```python

words = input().split()
count_dict = {}
for word in words:
    if word in count_dict:
        count_dict[word] += 1
    else:
        count_dict[word] = 1
print(count_dict)
```
